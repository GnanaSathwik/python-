##  File Handling and Operations
### Assignment 1: Reading a File

Write a function that reads the contents of a file named `sample.txt` and prints each line.

```python
def file_read(name):
    with open(name,'r') as file:
        line = file.readlines()
        for lines in line:
            print(lines)

file_read('sample.txt')
    
```

 ### Assignment 2: Writing to a File

Write a function that writes a list of strings to a file named `output.txt`, with each string on a new line.

```python
def file_write(lst,name):
    with open(name,'w') as file:
        for i in lst:
            file.write(f"{i} \n")
lst =['Hello Everyone', 'I am lord Shiva', 'I am one year old']
file_write(lst,'output.txt')
```

### Assignment 3: Copying a File

Write a function that copies the contents of a file named `source.txt` to a new file named `destination.txt`.

```python
def copy(source ,dest):
    with open(source ,'r') as file:
        content = file.read()
    
    with open(dest , 'a') as file:
        file.write(content)

copy('souce.txt', 'destination.txt')
```

### Assignment 5: Counting Words in a File

Write a function that reads the contents of a file named `source.txt` and returns the number of words in the file.

```python
def file_words(name):
    with open(name,'r') as file:
        lines = file.readlines()
        words = sum(len(line.split()) for line in lines)
    return words
file_words('souce.txt')
```

### Assignment 6: Finding and Replacing Text

Write a function that finds and replaces all occurrences of a given word in a file named `log.txt` with another word.

```python
def word_change(name,old ,new):
    with open(name,'r') as file:
        content = file.read()
    new = content.replace(old ,new)
    with open(name,'w') as file:
        file.write(new)

word_change('log.txt', 'dengey' , 'mingey')
```

### Assignment 7: Reading a File in Reverse

Write a function that reads the contents of a file named `sample.txt` and prints each line in reverse order.

```python
def rev(name):
    with open(name,'r') as file:
        lines = file.readlines()
    
    for line in reversed(lines):
        print(line.strip())

rev('sample.txt')

```

### Assignment 8: Counting Lines, Words, and Characters

Write a function that reads the contents of a file named `stats.txt` and returns the number of lines, words, and characters in the file.

```python
def count(name):
    with open(name,'r') as file:
        lines= file.readlines()
    
    num = len(lines)
    word = sum(len(line.split()) for line in lines)
    char = sum(len(line) for line in lines)

    return num , word, char
print(f"the number of lines ,words, and character { count('output.txt')}")
```

### Assignment 9: Merging Multiple Files

Write a function that merges the contents of multiple files into a single file named `merged.txt`.

```python
def merge(name1 , name2 , mix = 'merge.txt'):
    with open(name1,'r') as file:
        content = file.read()
    with open(name2,'r') as file:
        content1 = file.read()
    
    with open(mix,'a' ) as file:
        file.write(content)
        file.write("\n")
        file.write(content1)
        file.write("\n")

merge('log.txt' , 'sample.txt'  )
```

### Assignment 10: Splitting a Large File

Write a function that splits a large file named `large.txt` into smaller files of 100 lines each.

```python
def splitting(name , maxi):
    with open(name,'r') as file:
        lines = file.readlines()
    for i in range(0 , len(lines) ,maxi):
        with open(f"{name}_part{i //maxi + 1}.txt" ,'w') as sub:
            sub.writelines(lines[i:i+maxi])

splitting('large.txt',100)
```

### Assignment 11: Creating a Log File

Write a function that creates a log file named `activity.log` and writes log messages with timestamps.

```python
import datetime as dt
def log_msg(msg , name= 'activity.txt'):
    time = dt.datetime.now().isoformat()
    with open(name,'a') as file:
        file.write(f"({time}) : {msg}\n")

log_msg("I am creator of the world")
```

### Assignment 12: Binary File Operations

Write a function that reads a binary file named `image.bin` and writes its contents to another binary file named `copy_image.bin`.

```python
def copy_bin(source , dest = 'copy_image.bin'):
    with open(source ,'r') as s:
        with open(dest ,'w') as d:
            d.write(s.read())

copy_bin('image.bin')
```

### Assignment 13: CSV File Operations

Write a function that reads a CSV file named `data.csv` and prints its contents as a list of dictionaries.

```python
import csv

def read_csv(name):
    with open(name, 'r') as file:
        reader = csv.DictReader(file)
        data = list(reader)

    print(data)

read_csv('data.csv')
```

### Assignment 14: JSON File Operations

Write a function that reads a JSON file named `data.json` and prints its contents as a Python dictionary.

```python
import json

def read_json(name):
    with open(name, 'r') as file:
        reader = json.load(file)
    return reader

print(read_json('data.json'))
```

### Assignment 15: File Permission Handling

Write a function that attempts to read a file named `protected.txt` and handles any permission errors gracefully by printing an error message.

```python
def file_permission(name):
    try:
        with open(name,'r') as file:
            print(file.read())
    except PermissionError as e:
        print(f"Permission error: {e}")
    except FileNotFoundError as f:
        print(f"Error : {f}")

file_permission('protected.txt')
```
