### Assignment 1: Polymorphism with Methods

Create a base class named `Shape` with a method `area`. Create two derived classes `Circle` and `Square` that override the `area` method. Create a list of `Shape` objects and call the `area` method on each object to demonstrate polymorphism.

```python
from math import *

class Shape:
    def area(self):
        return "area of the shape"

class Circle(Shape):
    def __init__(self,rad):
        self.radius = rad
    def area(self):
        print(f"The area of the Circle with radius {self.radius} is: {pi*self.radius**2}")
class Square(Shape):
    def __init__(self,side):
        self.side = side
    def area(self):
        print(f"The area of the Square with side {self.side} is: {self.side**2}")

obj = [Circle(sqrt(1/pi) ), Square(7)]
for i in obj:
    i.area()
```

### Assignment 3: Abstract Base Class with Abstract Methods

Create an abstract base class named `Vehicle` with an abstract method `start_engine`. Create derived classes `Car` and `Bike` that implement the `start_engine` method. Create objects of the derived classes and call the `start_engine` method.

```python
from abc import ABC, abstractmethod
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass

class Car(Vehicle):
    def start_engine(self):
        print("Car Engine Started")

class Bike(Vehicle):
    def start_engine(self):
        print("Bike started")

BMW = Car()
duke = Bike()
BMW.start_engine()
duke.start_engine()
```

### Assignment 5: Encapsulation with Private Attributes

Create a class named `BankAccount` with private attributes `account_number` and `balance`. Add methods to deposit and withdraw money, and to check the balance. Ensure that the balance cannot be accessed directly.

```python
class Insufficientfunds(Exception):
    pass

class BankAccount:
    def __init__(self,acc,balance):
        self.__account_number = acc
        self.__balance = balance
        
    def check_balance(self):
        print(f"Your current balance Rs.{self.__balance}/-")

    def deposit(self,amt):
        self.__balance += amt
        print(f"Amount {amt} added to your bank account")

    def withdraw(self,amt):
        try:
            if(amt < self.__balance):
                self.__balance -= amt
            else:
                raise Insufficientfunds
        except Insufficientfunds:
            print(f"Insuffient Funds cannot with draw amt {amt} from ur bank account")

acc = BankAccount(1234567, 5000)
acc.withdraw(4000)
acc.check_balance()
acc.withdraw(3000)
acc.deposit(200000)
acc.check_balance()
acc.withdraw(100000)
acc.check_balance()
```

### Assignment 12: Abstract Properties

Create an abstract base class named `Appliance` with an abstract property `power`. Create two derived classes `WashingMachine` and `Refrigerator` that implement the `power` property. Create objects of the derived classes and access the `power` property.

```python
from abc import ABC,abstractmethod
class Appliance(ABC):
    @property
    @abstractmethod
    def power(self):
        pass

class WashingMachine(Appliance):
    @property
    def power(self):
        return 'Power consumed: 100W'
class Refrigerator(Appliance):
    @property
    def power(self):
        return "Power consumed 200W"
    
fridge = Refrigerator()
wash = WashingMachine()
print(fridge.power)
print(wash.power)
```

### Assignment 14: Polymorphism with Multiple Inheritance

Create a class named `Flyer` with a method `fly`. Create a class named `Swimmer` with a method `swim`. Create a class named `Superhero` that inherits from both `Flyer` and `Swimmer` and overrides both methods. Create an object of the `Superhero` class and call both methods.

```python
class Flyer:
    def fly(self):
        print("Flying")

class Swimmer:
    def swim(self):
        print("Swimming")

class Superhero(Flyer,Swimmer):
    def fly(self):
        print("Super Man is Flying")

    def swim(self):
        print("Super Man is Swimming")

hero = Superhero()
hero.fly()
hero.swim()
```

### Assignment 15: Abstract Methods and Multiple Inheritance

```python
from abc import ABC,abstractmethod
class Worker(ABC):
    @abstractmethod
    def work(self):
        pass

class Engineer(Worker):
    def work(self):
        return f"Engineer say they are the best"
    
class Doctor(Worker):
    def work(self):
        return f"Doctor say they are the best"

class Scientist(Engineer,Doctor): # whatever we write first in the argument that function is implemented(given prefernce)
    pass  ## Here in this case the Engineer is passed first so Engineers work method is implemented

man = Scientist()
print(man.work())
print(Doctor().work())
```
