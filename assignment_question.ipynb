**Question 5:** Write a Python program to calculate the factorial of a number.

```python
def fact(x):
    if x==0:
        return 1
    elif x<0:
        return "Invalid input"
    else:
        return x*fact(x-1)
num = int(input("enter the to find the factorial: "))
print(f"the factorial of {num} is {fact(num)}")
```

**Question 16:** Write a Python program to find the sum of the first n natural numbers.

```python
def sum(n):
    if n == 0:
        return 0
    elif n < 0:
        return " invalid input"
    else:
        return n+ sum(n-1)

n = int(input("Enter a number to find the sum of first n natural numbers: "))
print("the sum of first ",n," natural numbers is ", sum(n))
```

**Question 17:** Write a Python program to check if a year is a leap year.

```python
y = int(input("Enter a year to check whether it is a leap year or not: "))

if(y%4 == 0):
    if(y%100 == 0):
        if(y%400 == 0):
            print(f"{y} is a leap year")
        else:
            print(f"{y} is not a leap year")
    else:
        print(f"{y} is a leap year")
else:
    print(f"{y} is not a leap year")
```

**Question 18:** Write a Python program to reverse a string.

```python
str = input("enter a string to reverse the string:")
rev_str = ""
for i in range(len(str)-1,-1,-1):
    rev_str += str[i]
print(f"the reverse of the string {str} is {rev_str}")
```

**Question 19:** Write a Python program to check if a string is a palindrome.

```python
str = input("enter a string to check whether the string is a palindrome or not:")
rev_str = ""
for i in range(len(str)-1,-1,-1):
    rev_str += str[i]
if(str == rev_str):
    print(f"{str} is a palindrome")
else:
    print(f"{str} is not a palindrome")
```
