### Assignment 3: Data Cleaning

1. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Introduce some NaN values. Fill the NaN values with the mean of the respective columns.

```python
df = pd.DataFrame(np.random.randint(1,21,size=(5,3)))
indices = np.random.choice(15, size =3 , replace= False) 
rows , cols = np.unravel_index(indices= indices ,shape=(5,3))
for i,j in zip(rows, cols):
    df.iloc[i,j] = np.nan  
print("Array:")
display(df)
for i in df.columns:
    df[i] = df[i].fillna(df[i].mean())
print("\nFilled array:")
display(df)
```

2. Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers. Introduce some NaN values. Drop the rows with any NaN values.

### Assignment 5: Merging DataFrames

1. Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.

```python
df1 = pd.DataFrame({
    'Product': ["Laptop", "Mouse", "Shirt", "Jeans", "Apple"],
    "Category": ["Electronics", "Electronics", "Clothing", "Clothing", "Grocery"]
})

df2 =df1 = pd.DataFrame({
    'Product': ["Laptop", "Mouse", "Shirt", "Jeans", "Milk"],
    "Sales": [45000, 1200, 1500, 2500, 100]
}) 

print("Data Frame 1:")
display(df1)
print("Data Frame 2:")
display(df2)

print("Merged DataFrame")
pd.merge(df1,df2, on='Product')
```

2. Create two Pandas DataFrames with different columns. Concatenate the DataFrames along the rows and along the columns.

```python
df1 = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [23, 25, 22]
})
df2 = pd.DataFrame({
    "City": ["New York", "London", "Tokyo"],
    "Salary": [50000, 60000, 55000]
})
print("Data Frame 1:")
display(df1)
print("Data Frame 2:")
display(df2)

df_row = pd.concat([df1, df2], axis=0)
print("Row Concat")
display(df_row)

df_col = pd.concat([df1, df2], axis=1)
print("Column Concat")
display(df_col)

```

### Assignment 6: Time Series Analysis

1. Create a Pandas DataFrame with a datetime index and one column filled with random integers. Resample the DataFrame to compute the monthly mean of the values.

```python
df = pd.DataFrame({
    'Value': np.random.randint(1, 101, size=10)
},index = pd.to_datetime([
    "2025-01-05",
    "2025-01-12",
    "2025-01-20",
    "2025-02-03",
    "2025-02-18",
    "2025-03-01",
    "2025-03-15",
    "2025-03-25",
    "2025-04-10",
    "2025-04-22"
]))

print("Data Frame:")
display(df)

df.resample('ME').mean()
```

2. Create a Pandas DataFrame with a datetime index ranging from '2021-01-01' to '2021-12-31' and one column filled with random integers. Compute the rolling mean with a window of 7 days.

```python
r = pd.date_range('2021-01-01','2021-12-31', freq='D')
df = pd.DataFrame(r, columns=['Date'])
df['Value'] = np.random.randint(0,100,size = (len(r)))
df.set_index('Date',inplace=True)
print("Data Frame")
display(df)

roll = df.rolling(window=7).mean()
print("Rolling mean Data Frame")
display(roll.head(10))
```

### Assignment 8: Pivot Tables

1. Create a Pandas DataFrame with columns 'Date', 'Category', and 'Value'. Create a pivot table to compute the sum of 'Value' for each 'Category' by 'Date'.

```python
df = pd.DataFrame({
    'Date' : np.random.choice(
        pd.date_range('2024-01-01', '2024-12-31'),size= 10
    ),
    'Category': np.random.choice(['A','B','C'],size=10),
    'Value': np.random.randint(1,100,size = 10)
})

print("Data Frame:")
display(df)

df.groupby(['Category','Date'])['Value'].sum()
```
