### Assignment 9: Structured Arrays

1. Create a structured array with fields 'name' (string), 'age' (integer), and 'weight' (float). Add some data and sort the array by age.

```python
datatype = [('name', 'U10'), ('age','i4'), ('weight', 'f4')]
arr = np.array([
    ('Lord Shiva',2,10000),
    ('Sathwik',18,74.5)
],dtype=datatype)
print("Array:")
print(arr)
np.sort(arr , order = 'age')
print("\nSorted array:")
print(arr,"\n")

print(f"{'Name':<15} {'Age':<5} {'Weight':<10}") # here < is for left allignment and > is for right allignment
for row in arr:
    print(f"{row['name']:<15} {row['age']:<5} {row['weight']:<10}")
```

2. Create a structured array with fields 'x' and 'y' (both integers). Add some data and compute the Euclidean distance between each pair of points.

```python
datatype = [('x', 'i4'),('y', 'i4')]
arr = np.array([(1,2),(3,5),(7,18),(8,7)],dtype = datatype)
print(f"{'x':>5} {'y':>5}")
for row in arr:
    print(f"{row['x']:>5} {row['y']:>5}")

dist = np.sqrt((arr['x'][:, np.newaxis] - arr['x'])**2 + (arr['y'][:, np.newaxis] - arr['y'])**2)
print("Euclidean distances:")
print(dist)
```

### Assignment 10: Masked Arrays

1. Create a masked array of shape (4, 4) with random integers and mask the elements greater than 10. Compute the sum of the unmasked elements.

```python
import numpy.ma  as ma 

arr = np.random.randint(1,21,size = (4,4))
masked = ma.masked_greater(arr , 10 )
print("Array:")
print(arr)

print("\nMasked Array:")
print(masked)

print("\nSum")
print(masked.sum())
```

2. Create a masked array of shape (3, 3) with random integers and mask the diagonal elements. Replace the masked elements with the mean of the unmasked elements.

```python
import numpy.ma as ma

arr = np.random.randint(1,21,size=(3,3))
iden = np.eye(3 , dtype= bool)
masked = ma.masked_array(arr , mask= iden)
print("Array:")
print(arr)

print("\nMasked Array:")
print(masked)

mean = masked.mean()

masked = masked.filled(mean)
print("\nFilled Masked Array:")
print(masked)
```
