### Array Creation and Manipulation

Q1. Write a NumPy program to create a 3x3 matrix with values ranging from 2 to 10

```python
import numpy as np

arr = np.arange(2,11).reshape(3,3)
print(arr)
```

Q2. Write a NumPy program to create a 2D array with 1 on the border and 0 inside

```python
arr = np.ones(shape=(3,3))
arr[1:-1 ,1:-1 ] = 0
print(arr)
```

Q3. Write a NumPy program to add a border (filled with 0's) around an existing array.

```python
arr = np.arange(2,11).reshape(3,3)
print("Before adding")
print(arr)

print("After adding")
arr[0,:] = 0
arr[-1,:] = 0
arr[:,0] = 0
arr[:,-1] = 0
print(arr)

```

Q4.Write a NumPy program to create an 8x8 matrix and fill it with a checkerboard pattern

```python
checker = np.zeros(shape=(8,8))
print(checker)
checker[::2 , 1::2] = 1
checker[1::2, ::2] = 1
print("\nChecker Board")
print(checker)
```

### Indexing and Selection 

Q15.Write a NumPy program to test whether each element of a 1-D array is also present in a second array.

```python
import numpy as np
arr1 = np.array([1,2,3,4,5,6,7])
arr2 = np.arange(1,21).reshape(4,5)
r = np.isin(arr1,arr2)
print(r)
```

Q16.Write a NumPy program to get the unique elements of an array.

```python
arr = np.random.randint(1,21,size=(5,5))
print(arr)
print(np.unique_values(arr))
```

Q17.Write a NumPy program to find the set difference between two arrays. The set difference will return sorted, distinct values in array1 that are not in arra

```python
arr1 = np.array([1,2,3,4,5,6,7])
arr2 = np.array([1,7,18,8,45])
print(np.setdiff1d(arr1,arr2))
```

Q18.Write a NumPy program to find the set exclusive-or of two arrays. Set exclusive-or will return sorted, distinct values in only one (not both) of the input arrays.

```python
arr1 = np.array([1,2,3,4,5,6,7])
arr2 = np.array([1,7,18,8,45])
print(np.setxor1d(arr1,arr2))
```

Q19.Write a NumPy program to find the union of two arrays. Union will return a unique, sorted array of values in each of the two input arrays

```python
arr1 = np.array([1,2,3,4,5,6,7])
arr2 = np.array([1,7,18,8,45])
print(np.union1d(arr1,arr2))
```

Q20.Write a NumPy program to get the values and indices of the elements that are bigger than 10 in a given array

```python
arr = np.random.randint(1,21,size=20)
a = arr[arr>10]
indices = np.where(arr>10)[0]
print([(int(i), int(v)) for i, v in zip(indices, a)])
```

Q24.Write a NumPy program to select indices satisfying multiple conditions in a NumPy array.

```python
arr = np.random.randint(1,51,size = 20)
indices = np.where((arr > 30) | (arr<10))[0]
print(arr,"\n")

for i,j in zip(indices, arr[indices]):
    print(i,j)
```
