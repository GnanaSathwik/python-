## Mathematics and Linear Algebra

Q12.Write a NumPy program to find the indices of the maximum and minimum values along the given axis of an array.

```python
arr = np.random.randint(1,51,size=(6,5))
print(arr,"\n")
max_idx = np.argmax(arr,axis=1) # Give row wise
min_idx = np.argmin(arr,axis=1) # Give row wise
print("Row wise:")
print(max_idx)
print(min_idx)
print("\nMax Values")
for i,idx in enumerate(max_idx):
    print(arr[i,idx],end=" ")
print("\nMin Values")
for i,idx in enumerate(min_idx):
    print(arr[i,idx],end=" ")
print()
max_idx = np.argmax(arr,axis=0) 
min_idx = np.argmin(arr,axis=0)
print("\nColumn Wise: ")
print(max_idx)
print(min_idx)
print("\nMax Values")
for i,idx in enumerate(max_idx):
    print(arr[idx,i],end=" ")

print("\nMin Values")
for i,idx in enumerate(min_idx):
    print(arr[idx,i],end=" ")

```

Q18.Write a NumPy program to remove all rows in a NumPy array that contain non-numeric values

```python
arr = np.array([[1,8,99],
                [4,5,np.nan],
                [9,10,np.inf],
                [7,18,45]])

arr = arr[np.isfinite(arr).all(axis=1)]
print(arr)
```
