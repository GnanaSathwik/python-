### Assignment 7: Handling Transactions

1. Write a Python function to insert multiple employees into the `employees` table in a single transaction. Ensure that if any insertion fails, none of the insertions are committed.

```python
#  ("The updated version of the insert bunch")

def insert_bunch(db ,table,  lst):
    con = sql.connect(db)
    cursor = con.cursor()
    try:
        cursor.executemany(f'''
Insert Into {table} (name,age,department)
        Values (?,?,?)
''',  lst)
        
        con.commit()
        print("All records inserted succesfully")
    except  Exception as e:
        con.rollback()
        print("Succesfully rolled back")
        print(e)
    finally:
        con.close()


records = [
    ( 'Frank', 40, 'Finance'),
    ( 'Grace', 29, 'Engineering'),
    ( 'Hannah', 35, 'Marketing'),
    ( 'Ivan', 38, 'Sales'),
    ( 5, 'Jack', 45, 'HR') ] 
insert_bunch('test.db','employees' ,records)
```

### Assignment 8: Creating Relationships

1. Create a new table named `departments` with columns `id` (integer) and `name` (text).

```python
con = sql.connect('test.db')
cursor = con.cursor()
cursor.execute('''
Create Table if not exists departments (
               id Integer Primary Key,
               name  Text Not null
               )

''')
con.commit()
con.close()
print("Department table create succesfully")
```

2. Modify the `employees` table to include a foreign key referencing the `id` column in the `departments` table.

```python
con = sql.connect('test.db')
cursor = con.cursor()
cursor.executescript('''
Pragma foreign_keys = off;
               Begin Transaction;
               Alter table employees Rename to old;
               Create Table employees(
               id Integer Primary Key,
               name Text Not null,
               age Integer,
               department Text,
               dept_id Integer,
               Foreign Key (dept_id) References departments(id)                               
            );
            
               Insert into employees (id,name,age,department)
               select  id , name , age, department from old ;
               Drop Table old;
               Commit;
               Pragma foreign_key  = on
''')

con.commit()
con.close()
print("Table employee is updated")
```

3. Write a Python function to insert data into both the `departments` and `employees` tables, ensuring referential integrity.

```python
def insert_foreign(db,id,name,age, dept, dept_id):
    con = sql.connect('test.db')
    cursor = con.cursor()
    try:
        cursor.execute('''
    Insert into departments(id,name)
                       values(?,?)
''', (dept_id, dept ))
        cursor.execute('''
Insert into employees (id, name ,age, department, dept_id)
                       values (?,?,?,?,?)                        
''', (id, name, age, dept,dept_id))
        con.commit()
    except Exception as e:
        con.rollback()
        print("Succesully roll backed")
        print(e)
    
    finally:
        con.close()

insert_foreign('test.db',18, 'MS Dhoni', 43 , 'Cricket', 10)
```
