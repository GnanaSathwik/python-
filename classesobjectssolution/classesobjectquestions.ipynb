### Assignment 11: Abstract Base Class

Create an abstract base class named `Shape` with an abstract method `area`. Create derived classes `Circle` and `Square` that implement the `area` method. Create objects of the derived classes and call the `area` method.

```python
from abc import ABC, abstractmethod
from math import *
class Shape(ABC):
    @abstractmethod
    def area():
        pass

class Circle(Shape):
    def __init__(self,r):
        self.radius = r
    def area(self):
        return pi*self.radius*self.radius

class Square(Shape):
    def __init__(self,side):
        self.side = side
    def area(self):
        return self.side*self.side
    
a = Circle(14)
b = Square(6)
print(a.area())
print(b.area())
```

### Assignment 12: Operator Overloading

Create a class named `Vector` with attributes `x` and `y`. Overload the `+` operator to add two `Vector` objects. Create objects of the class and test the operator overloading.

```python
class Vector:
    def __init__(self,x,y):
        self.x =x
        self.y = y
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
    def __str__(self):
        return f"Vector ({self.x},{self.y})"

obj1 = Vector(5,6)
obj2 = Vector(2,12)
print(obj1.__add__(obj2))
```

### Assignment 13: Class with Custom Exception

Create a custom exception named `InsufficientBalanceError`. In the `BankAccount` class, raise this exception when a withdrawal amount is greater than the balance. Handle the exception and print an appropriate message.

```python
class InsufficientBalanceError(Exception):
    pass

class BankAccount:
    def __init__(self , account_number , balance):
        self.__account_number = account_number
        self.__balance = balance
    def deposit(self,amount):
        self.__balance += amount
        print(f"Amount {amount}/- credited to your Bank Account")
    def withdraw(self, amt):
        try:

            if(amt > self.__balance):
                raise InsufficientBalanceError(f"Insufficient Funds in your bank account")
            else:
                self.__balance -= amt
                print(f"Amount {amt}/- debited from your Bank Account")
        except InsufficientBalanceError as e:
            print(e)
    def check_balance(self):
        print(f"Balance Amount: {self.__balance}")
acc = BankAccount(123456789 , 5000)
acc.withdraw(2000)
acc.withdraw(4000)
```

### Assignment 14: Class with Context Manager

Create a class named `FileManager` that implements the context manager protocol to open and close a file. Use this class to read the contents of a file.

```python
class FileManager:
    def __init__(self,name,mode):
        self.name = name
        self.mode = mode
        self.file = None
    def __enter__(self):
        self.file= open(self.name, self.mode)
        return self.file

    def __exit__(self,exc_type,exc_value,traceback):
        self.file.close()

with FileManager('sample.txt','w') as file:
    file.write("Hi I am Lord Shiva")
```

### Assignment 15: Chaining Methods

Create a class named `Calculator` with methods to add, subtract, multiply, and divide. Each method should return the object itself to allow method chaining. Create an object and chain multiple method calls.

```python
class Calculator:
    def __init__(self,value = 0):
        self.value = value
    
    def add(self,val):
        self.value+=val
        return self
    
    def subtract(self,val):
        self.value -= val
        return self
    
    def multiply(self, val):
        self.value*= val
        return self
    
    def divide(self, val):
        try:
            self.value /= val
        except ZeroDivisionError as e:
            print("Cannot divide by zero")
        return self

cal = Calculator()
cal.add(10).multiply(2).subtract(10).divide(0)
print(cal.value)
            
```
